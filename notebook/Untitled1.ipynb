```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


df = pd.read_csv("train.csv")

print("\n--- Missing Values ---")
print(df.isnull().sum())

print("\n--- Survival Count ---")
print(df["Survived"].value_counts())

survivors = df[df["Survived"] == 1]
non_survivors = df[df["Survived"] == 0]

print("\nNumber of Survivors:", len(survivors))
print("Number of Non-Survivors:", len(non_survivors))

print("\n--- Passengers by Sex ---")
print(df["Sex"].value_counts())

print("\n--- Passengers by Class ---")
print(df["Pclass"].value_counts().sort_index())

print("\n--- Passengers by Sex and Class ---")
print(df.groupby(["Sex", "Pclass"]).size())

ages = df["Age"].dropna().to_numpy()
fares = df["Fare"].dropna().to_numpy()

print("\n--- NumPy Calculations ---")

print("Average Age:", np.mean(ages))
print("Median Age:", np.median(ages))

print("Average Fare:", np.mean(fares))
print("Median Fare:", np.median(fares))

print("\n--- Survival by Sex ---")
print(df.groupby("Sex")["Survived"].sum())

print("\n--- Survival by Passenger Class ---")
print(df.groupby("Pclass")["Survived"].sum())

print("\n--- Survival by Sex and Class ---")
print(df.groupby(["Sex", "Pclass"])["Survived"].sum())

print("\n--- Top 10 Highest Fares ---")

top_10_fares = df.sort_values(
    by="Fare", ascending=False
).head(10)

print(top_10_fares[["Name", "Fare", "Pclass"]])

print("\n--- Female Passengers ---")
print(df[df["Sex"] == "female"].head())

print("\n--- First Class Passengers ---")
print(df[df["Pclass"] == 1].head())

print("\n--- Passengers Who Survived ---")
print(df[df["Survived"] == 1].head())


survival_counts = df["Survived"].value_counts().sort_index()

plt.figure(figsize=(7, 5))

plt.bar(
    ["Not Survived", "Survived"],
    survival_counts.values
)

plt.title("Titanic Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.show()

plt.figure(figsize=(7, 5))

plt.hist(df["Age"].dropna(), bins=20)

plt.title("Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.show()

print("\n--- 3 Observations ---")

print("1. More passengers died than survived.")

print("2. Female passengers had a higher survival count "
      "compared to male passengers.")

print("3. First-class passengers had better survival "
      "numbers compared to other passenger classes.")

print("\n========== PROJECT COMPLETED ==========")
```
